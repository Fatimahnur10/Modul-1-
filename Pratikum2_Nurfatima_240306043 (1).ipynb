## 1. Variabel Linguistik CPU Utilization

Variabel linguistik yang digunakan adalah CPU Utilization dengan satuan persen (%).

Semesta pembicaraan:
U = [0, 100]%

Himpunan fuzzy yang digunakan terdiri dari:
1. Rendah
2. Normal
3. Tinggi

Himpunan Rendah menggunakan bentuk left shoulder.
Himpunan Normal menggunakan bentuk triangle.
Himpunan Tinggi menggunakan bentuk right shoulder.

## 2. Fungsi Keanggotaan Rendah

Himpunan Rendah memiliki nilai keanggotaan penuh pada CPU 20% atau kurang.
Pada rentang 20% sampai 40%, nilai keanggotaan menurun secara linear.
Pada CPU 40% atau lebih, nilai keanggotaan adalah 0.

Rumus:


### Fungsi Keanggotaan Rendah
μ_Rendah(x) =

\begin{cases}
1, & x \leq 20\\
\dfrac{40-x}{20}, & 20 < x < 40\\
0, & x \geq 40
\end{cases}



Contoh perhitungan

Misalnya CPU = 35%:

$$ \mu_{Rendah}(35) =\frac{40-35}{20} =\frac{5}{20} =0.25 $$

Jadi derajat keanggotaan CPU 35% terhadap Rendah = 0.25.

## 3. Fungsi Keanggotaan Normal

Himpunan Normal menggunakan bentuk segitiga dengan titik:
(30, 0), (50, 1), dan (70, 0).

Rumus:

μ_Normal(x) =




\begin{cases}
0, & x \leq 30 \text{ atau } x \geq 70\\
\dfrac{x-30}{20}, & 30 < x \leq 50\\
\dfrac{70-x}{20}, & 50 < x < 70
\end{cases}



Contoh CPU = 35%:

$$ \mu_{Normal}(35) =\frac{35-30}{20} =\frac{5}{20} =0.25 $$

Jadi Normal = 0.25.

## 4. Fungsi Keanggotaan Tinggi

Himpunan Tinggi memiliki nilai keanggotaan 0 pada CPU 60% atau kurang.
Pada rentang 60% sampai 80%, nilai keanggotaan meningkat secara linear.
Pada CPU 80% atau lebih, nilai keanggotaan adalah 1.

Rumus:

μ_Tinggi(x) =

\begin{cases}
0, & x \leq 60\\
\dfrac{x-60}{20}, & 60 < x < 80\\
1, & x \geq 80
\end{cases}



Contoh CPU = 75%:

$$ \mu_{Tinggi}(75) =\frac{75-60}{20} =\frac{15}{20} =0.75 $$

Jadi Tinggi = 0.75.

Ketiga bentuk fungsi ini mengikuti spesifikasi tugas CPU pada modul.

## Kesimpulan Perhitungan Manual

Berdasarkan perhitungan manual, setiap nilai CPU dapat memiliki derajat keanggotaan yang berbeda pada himpunan Rendah, Normal, dan Tinggi. Pada nilai yang berada di daerah overlap, seperti 35% dan 65%, satu input dapat memiliki derajat keanggotaan pada dua himpunan fuzzy sekaligus. Hasil perhitungan manual tersebut kemudian dibandingkan dengan hasil perhitungan menggunakan program Python.

```python
import numpy as np
import matplotlib.pyplot as plt

# Fungsi keanggotaan Rendah
def mf_rendah(x):
    if x <= 20:
        return 1
    elif x < 40:
        return (40 - x) / 20
    else:
        return 0


# Fungsi keanggotaan Normal
def mf_normal(x):
    if x <= 30 or x >= 70:
        return 0
    elif x <= 50:
        return (x - 30) / 20
    else:
        return (70 - x) / 20


# Fungsi keanggotaan Tinggi
def mf_tinggi(x):
    if x <= 60:
        return 0
    elif x < 80:
        return (x - 60) / 20
    else:
        return 1
```

```python
variabel_linguistik_cpu = {
    "Rendah": mf_rendah,
    "Normal": mf_normal,
    "Tinggi": mf_tinggi
}
```

```python
def fuzzifikasi_cpu(x):
    hasil = {}

    for label, fungsi in variabel_linguistik_cpu.items():
        hasil[label] = fungsi(x)

    return hasil
```

```python
x = 35

hasil = fuzzifikasi_cpu(x)

print("CPU Utilization =", x, "%")
print("Rendah =", hasil["Rendah"])
print("Normal =", hasil["Normal"])
print("Tinggi =", hasil["Tinggi"])
```

```python
data_uji = [10, 35, 50, 65, 75, 95]

print("CPU (%) | Rendah | Normal | Tinggi")
print("------------------------------------")

for x in data_uji:
    hasil = fuzzifikasi_cpu(x)

    print(
        f"{x:7} | "
        f"{hasil['Rendah']:.2f}   | "
        f"{hasil['Normal']:.2f}   | "
        f"{hasil['Tinggi']:.2f}"
    )
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Fungsi keanggotaan CPU Rendah
def mf_rendah(x):
    if x <= 20:
        return 1
    elif x < 40:
        return (40 - x) / 20
    else:
        return 0

# Fungsi keanggotaan CPU Normal
def mf_normal(x):
    if x <= 30 or x >= 70:
        return 0
    elif x <= 50:
        return (x - 30) / 20
    else:
        return (70 - x) / 20

# Fungsi keanggotaan CPU Tinggi
def mf_tinggi(x):
    if x <= 60:
        return 0
    elif x < 80:
        return (x - 60) / 20
    else:
        return 1

# Membuat nilai sumbu X
x = np.linspace(0, 100, 500)

# Menghitung derajat keanggotaan
y_rendah = np.array([mf_rendah(i) for i in x])
y_normal = np.array([mf_normal(i) for i in x])
y_tinggi = np.array([mf_tinggi(i) for i in x])

# Membuat grafik
plt.figure(figsize=(10, 6))

plt.plot(x, y_rendah, label="Rendah")
plt.plot(x, y_normal, label="Normal")
plt.plot(x, y_tinggi, label="Tinggi")

plt.title("Fungsi Keanggotaan CPU Utilization")
plt.xlabel("CPU Utilization (%)")
plt.ylabel("Derajat Keanggotaan")

plt.ylim(-0.05, 1.05)
plt.grid(True)
plt.legend()

# Menyimpan grafik
plt.savefig("variabel_linguistik_cpu.png", dpi=300, bbox_inches="tight")

plt.show()
```

Mengapa kurva fungsi keanggotaan yang bersebelahan harus saling tumpang tindih (overlap)? Apa konsekuensinya pada sistem inferensi jika terdapat celah (gap) kosong di antara dua kurva?


Jawaban:

Kurva fungsi keanggotaan yang bersebelahan perlu saling tumpang tindih (overlap) agar setiap nilai input dapat memiliki derajat keanggotaan pada satu atau lebih himpunan fuzzy. Dengan adanya overlap, perubahan nilai input dapat berlangsung secara bertahap dari satu kategori ke kategori lainnya sehingga tidak terjadi perubahan keputusan yang terlalu tiba-tiba.

Jika terdapat celah (gap) antara dua kurva, maka terdapat nilai input yang tidak memiliki derajat keanggotaan pada kedua himpunan tersebut. Kondisi ini dapat menyebabkan tidak ada aturan fuzzy yang aktif untuk nilai input tersebut, sehingga proses inferensi dapat menghasilkan keputusan yang kurang tepat atau bahkan tidak menghasilkan keluaran.
